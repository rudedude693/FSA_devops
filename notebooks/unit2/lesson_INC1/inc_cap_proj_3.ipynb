# Task: Analyze Data #

```python
# Univariate Statistical Analysis
import matplotlib.pyplot as plt
data_to_analyze = ['num_physician_office_visits',
                   'num_non_physician_office_visits',
                   'num_outpatient_physician_office_visits',
                   'num_outpatient_non_physician_office_visits',
                   'num_emergency_visits',
                   'gender',
                   'married',
                   'num_school_years',
                   'income',
                   'employed',
                   'has_priv_insurance',
                   'has_medicaid_coverage'
                   ]
for col in data_to_analyze:
    print(f'For column: {col}')
    std_dev = nmses_df[col].std()
    mean = nmses_df[col].mean()
    median = nmses_df[col].median()
    #mode = nmses_df[col].mode()
    print(f'std: {std_dev}')
    print(f'mean: {mean}')
    print(f'median: {median}')
    #print(f'mode: {mode}')
    sns.histplot(data=nmses_df, x=col)
    plt.axvline(x=std_dev,label='std')
    plt.axvline(x=mean,label='mean')
    plt.axvline(x=median,label='median')
    #plt.axvline(x=mode,label='mode')
    plt.show()
    
```

## Result ##
Inital plots of central tendency show data shapes of the binary data points (gender, married, etc.) don't make much sense to plot with std dev, mean, median.
The others tend to have a negative skew, especially the visits. Since many reported visits are done in urgent or emergency care they skew heavily towards 0.
Even normal visits are typically on the low end, though the curve was more gradual.

```python
import numpy as np
bins = range(60,110,5)
sns.histplot(data=nmses_df, x='age', hue='gender', bins=bins, legend=False)
plt.legend(title='Gender', loc='upper right', labels=['Male', 'Female'])
plt.xticks(bins)
plt.show()

fig = sns.countplot(data=nmses_df, x='gender', hue='self_reported_health_factor', legend=False)
#print(nmses_df['self_reported_health_factor'].value_counts())
plt.legend(title='Health', loc='upper right', labels=['Average', 'Poor', 'Excellent'])
fig.set_xticklabels(['Male','Female'])
plt.show()

#print(nmses_df['income'].max())

#bins = np.logspace(0,int(nmses_df['income'].max()),num=11, base=10)
#bins=range(0,int(nmses_df['income'].max()),35000)
fig = sns.histplot(data=nmses_df, x='income', hue='gender', legend=False, log_scale=True, bins=20, multiple='dodge')
plt.legend(title='Gender', loc='upper right', labels=['Male', 'Female'])
fig.set_xlabel('Income (log)')
plt.show()

fig = sns.histplot(data=nmses_df, x='income', hue='region', legend=False, log_scale=True, bins=20, multiple='dodge')
plt.legend(title='region', loc='upper right', labels=nmses_df['region'].unique())
fig.set_xlabel('Income (log)')
plt.show()

fig = sns.barplot(data=nmses_df, x='income', hue='age')
#plt.legend(title='region', loc='upper right', labels=nmses_df['region'].unique())
fig.set_xlabel('Income')
fig.set_ylabel('Age')
fig.set_yticks(range(120,50,5))
fig.set_yticklabels(range(110,60,5))
plt.show()
```
